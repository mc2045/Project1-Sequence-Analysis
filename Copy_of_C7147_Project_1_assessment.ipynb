## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Validate read

    Args:
        sequence: A sequence string to validate, which may contain invalid characters.

    Returns:
        A dictionary with the keys sequence, valid, and errors.
    """
    # Clean sequence. White space removal not required.
    sequence = sequence.upper()
    # Initial values assume no error.
    errors = []
    valid = True
    # Test length of sequence is the same as length of reference.
    if len(sequence) != len(reference):
        # Add error to the list.
        errors.append("wrong_length")
        valid = False
    # Test sequence contains valid bases.
    if not all(base in "ATGC" for base in sequence):
        # Add error to the list.
        errors.append("invalid_characters")
        valid = False
    # Return the dictionary as specified.
    return {"sequence":sequence, "valid":valid, "errors":errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Find mutations in seuquence using reference.
    Iterates each character in  sequence and compares with corresponding character in reference.
    Returns a list of dictionaries describing mismatches."""
    # Intialise output.
    output = []
    # Visit each character in sequence.
    for i in range(len(sequence)):
        character = sequence[i:i+1]
        # Get corresponding reference character.
        ref_character = reference[i:i+1]
        # Compare with a reference character.
        if character != ref_character:
            # Append details of invlaid character.
            output.append({"position": i+1, "reference_base": ref_character, "read_base": character})
    return output

````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Analyse NGS reads.
    Iterates each read. Validating each read and where reads are valid finding mutations.
    Returns a new list of dictionaries containing the analysed read with cleaned sequences and additional new information
    on status (valid/invalid), errors and mutations"""

    # Initialise output.
    output = []
    # Iterate reads.
    for read in reads:
        # Validate read
        validate_result = validate_read(read["sequence"], reference)
        # If it is valid, find mutations.
        if validate_result["valid"]:
            mutations = find_mutations(validate_result["sequence"], reference)
        else:
            # make sure to return an empty list for invalid reads.
            mutations = []
        # Build output line.
        output_line = read.copy()
        # Replace sequence with clean one.
        output_line["sequence"] = validate_result["sequence"]
        # Map valid (true/false) to status (valid/invalid).
        output_line["status"] = "valid" if validate_result["valid"] else "invalid"
        # Add errors.
        output_line["errors"] = validate_result["errors"]
        # Add mutations.
        output_line["mutations"] = mutations
        # Append output line to the output.
        output.append(output_line)
    return output
````

````python
# Write your code here.

def get_sample_set(analysed_reads, sample):
    """get a sample set from analysed reads."""
    return [read for read in analysed_reads if read["sample"] == sample]

def get_valid_reads(sample_set):
    """counts reads marked valid."""
    count = 0
    for read in sample_set:
        if(read["status"] == "valid"): count += 1
    return count


def get_invalid_reads(sample_set):
    """counts reads marked invalid."""
    count = 0
    for read in sample_set:
        if(read["status"] == "invalid"): count += 1
    return count

def get_mismatch_reads(sample_set):
    """counts reads containing at least one mismatch."""
    count = 0
    for read in sample_set:
        if len(read["mutations"]) > 0:
            count += 1
    return count

def get_mismatch_events(sample_set):
    """counts all mismatch events across all the reads."""
    count = 0
    for read in sample_set:
        count += len(read["mutations"])
    return count

def get_substitution_counts(sample_set):
    """returns counts for each observed substitution type."""
    output = {}
    for read in sample_set:
        for mutation in read["mutations"]:
            key = mutation["reference_base"] + ">" + mutation["read_base"]
            if key in output:
                output[key] += 1
            else:
                output[key] = 1
    # Sort the output alphabetically
    sorted_output = {}
    for key_value in sorted(output.items(), key=lambda key_value: key_value[0]):
      sorted_output[key_value[0]] = key_value[1]
    return sorted_output

def get_position_counts(sample_set):
    """returns mismatch counts by reference position."""
    output = {}
    for read in sample_set:
        for mutation in read["mutations"]:
            key = mutation["position"]
            if key in output:
                output[key] += 1
            else:
                output[key] = 1
    # Sort the output by position
    sorted_output = {}
    for key_value in sorted(output.items(), key=lambda key_value: key_value[0]):
      sorted_output[key_value[0]] = key_value[1]
    return sorted_output

def get_top_positions(positions):
    """returns up to three positions ranked by count, resolving ties by earlier position."""
    # Sort the postions by Value DESCENDING and Key ASCENDING, then slice the first three records to get the top 3.
    top_positions = [key_value[0] for key_value in sorted(positions.items(), key=lambda key_value: (-key_value[1],key_value[0]))[:3]]
    return top_positions

def get_unique_samples(analysed_reads):
    """returns unique sample-group pairs sorted by sample name."""
    unique_samples = sorted(
        {(read["sample"], read["group"]) for read in analysed_reads },
        key=lambda row: row[0]
    )
    return unique_samples

def summarise_samples(analysed_reads):
    """analysed read-input, returned sample-summary dictionary, and printed summaries."""

    #Initialise the output
    output = {}

    # Get a unique list of samples, note that groups and sample seem to have a 1:1 relationship, so get both
    unique_samples = get_unique_samples(analysed_reads)

    check_samples = {(read["sample"]) for read in analysed_reads}

    # if(len(check_samples) == 0): raise Exception("No samples found")

    if(len(check_samples) != len(unique_samples)): raise Exception("Mismatch between groups and samples detected")

    for sample, group in unique_samples:
        sample_set = get_sample_set(analysed_reads, sample)
        output[sample] = {
            "group": group,
            "valid_reads": get_valid_reads(sample_set),
            "invalid_reads": get_invalid_reads(sample_set),
            "mismatch_reads": get_mismatch_reads(sample_set),
            "mismatch_percentage": 0, # Placeholder
            "mismatch_events": get_mismatch_events(sample_set),
            "substitution_counts": get_substitution_counts(sample_set),
            "position_counts": get_position_counts(sample_set),
            "top_positions": []
        }
        # Calculate mismatch percentage
        output[sample]["mismatch_percentage"] = output[sample]["mismatch_reads"] / output[sample]["valid_reads"] * 100 if output[sample]["valid_reads"] > 0 else None

        output[sample]["top_positions"] = get_top_positions(output[sample]["position_counts"])

    # output table


    print("Table of summarised results\n")
    print(f"{'Sample':<15}|{'Valid Reads':<15}|{'Invalid Reads':<15}|{'Percentage':<15}|{'Mismatch Events':<15}|{'Top Positions':<15}")
    print("-" * 94)
    for sample in output:
        result = output[sample]
        top_positions_str = ", ".join(str(x) for x in result["top_positions"])
        print(f"{sample:<15}|{result["valid_reads"]:<15}|{result["invalid_reads"]:<15}|{result["mismatch_percentage"]:<15.4f}|"
            + f"{result["mismatch_events"]:<15}|{top_positions_str:<15}")

    print("\nTable of substitution mutations\n")
    # How may rows will there be so we can make the column dividers pretty
    maxrows = 0
    for sample in output:
        result = output[sample]
        maxrows = max(len(result["substitution_counts"]),maxrows)
    cols = ""
    rule = ""
    first = True
    # Headers
    for sample in output:
        result = output[sample]
        if not first:
          cols += '|'
          rule += '-'
        cols += f"{sample:<20}"
        rule += "-" * 20
        first = False
    print(cols)
    print(rule)
    # Rows
    for i in range(maxrows):
        row = ""
        first = True
        for sample in output:
            result = output[sample]
            substitution_counts = list(result["substitution_counts"].items())
            if not first: row += '|'
            first = False
            if(i < len(substitution_counts)):
                # Need to calculate the percentage as it is not in output
                percent = substitution_counts[i][1] / result["valid_reads"] * 100
                rowvalue = f"{substitution_counts[i][0]}: {substitution_counts[i][1]} ({percent:.2f}%)"
            else:
                rowvalue = ""
            row += f"{rowvalue:<20}"
        print(row)
    # done
    return output
````

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `control_group` and `treated_group` are strings

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.

- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 💻 Enter your code here

````python


def consolidate_position_counts (sample_summary):
    """Consolidates position counts across all samples into a single list."""
    output = {}
    for sample, sample_data in sample_summary.items():
        position_counts = sample_data["position_counts"]
        #print("position_counts",position_counts)
        for position in position_counts:
            if position in output:
                output[position] += position_counts[position]
            else:
                output[position] = position_counts[position]

            #print("output",output)

    return output

def get_percentage_lists (sample_summary, top_positions, control_group, treated_group):
    """For each group, and for each top postion, this gets a lists of
    postion counts as a percentage of the valid reads for that group """
    #Initialise the percentage lists for each group
    percentage_lists = {control_group:[], treated_group:[]}
    for sample, sample_data in sample_summary.items():
        group = sample_data["group"]
        valid_reads = sample_data["valid_reads"]
        if(group != control_group and group != treated_group): continue # Just in case
        position_counts = sample_data["position_counts"]
        # Set up the mismatch percentages array
        mismatch_percentages = [0] * len(top_positions)
        # Calculate mismatch percentages
        for i in range(len(top_positions)):
            if top_positions[i] in position_counts and valid_reads > 0:
                mismatch_percentages[i] = position_counts[top_positions[i]] / valid_reads * 100
            else:
                mismatch_percentages[i] = 0

        percentage_lists[group].append(mismatch_percentages)

    return percentage_lists

# Can only achieve the specification by adding analysed_reads to input parameters
def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Main compare positions function
    1) Consolidates position counts across all samples into a single list.
    2) From that list gets the top 3 positions
    3) From the top 3 positions calls get_percentage_lists
    4) Performs statistical analysis on the list of percentages at each postion
    across the two groups (control and treated)
    5) Outputs the table of statistics covering each sample
    6) Returns a dictionary containing the statisics for each sample
    """

    position_counts = consolidate_position_counts(sample_summary)
    #print(position_counts)
    top_positions = get_top_positions(position_counts)
    #print(top_positions)

    percentage_lists = get_percentage_lists(sample_summary,top_positions, control_group, treated_group)

    #print(percentage_lists)

    # Convert to numpy arrays to aid slicing

    percentage_arrays = {control_group:np.array(percentage_lists[control_group]),
                         treated_group:np.array(percentage_lists[treated_group])}

    #print(percentage_arrays)

    # Results table header

    print("Table of statistics by mutation position (two-sided Welch's t-test)\n")
    print(f"{'Position':<15}|{'Control Count':<15}|{'Treated Count':<15}|{'Control Mean':<15}|{'Treated Mean':<15}|{'Difference':<15}|" +
            f"{'Statistic':<15}|{'p-value':<15}")
    print("-" * 126)

    # Loop through the top positions and call the statistics function for each

    output = []
    for position in range(0,len(top_positions)):
        # Slice the respective columns from the percentage arrays and compute the statistics
        control_slice = percentage_arrays[control_group][:, position]
        treated_slice = percentage_arrays[treated_group][:, position]
        stats = calculate_statistics(control_slice,treated_slice)
        # Get means and difference between the means
        control_mean = np.mean(control_slice)
        treated_mean = np.mean(treated_slice)
        difference = treated_mean - control_mean
        # Append the results to the output
        result = {
                    "position": top_positions[position],
                    "control_n": len(control_slice),
                    "treated_n": len(treated_slice),
                    "control_mean": float(control_mean) if control_mean != None else None,
                    "treated_mean": float(treated_mean) if treated_mean != None else None,
                    "difference": float(difference) if difference != None else None,
                    "statistic": float(stats["statistic"]) if stats["statistic"] != None else None,
                    "p_value": float(stats["p_value"]) if stats["p_value"] != None else None
                }

        output.append(result)
        # print the output
        control_mean = f"{result["control_mean"]:<15.4f}" if result["control_mean"] != None else f"{'N/A':<15}"
        treated_mean = f"{result["treated_mean"]:<15.4f}" if result["treated_mean"] != None else f"{'N/A':<15}"
        difference = f"{result["difference"]:<15.4f}" if result["difference"] != None else f"{'N/A':<15}"
        statistic = f"{result["statistic"]:<15.4f}" if result["statistic"] != None else f"{'N/A':<15}"
        p_value = f"{result["p_value"]:<15.4f}" if result["p_value"] != None else f"{'N/A':<15}"
        print(f"{result["position"]:<15}|{result["control_n"]:<15}|{result["treated_n"]:<15}|{control_mean}|"
            + f"{treated_mean}|{difference}|{statistic}|{p_value}")

    return output
````
